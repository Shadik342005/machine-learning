```python
def f(x, y):
    return np.sin(np.sqrt(x**2+y**2))
```

```python
r = np.linspace(25, 55, )
theta = np.linspace(-0.9*np.pi, 0.8*np.pi, 40)

r,theta = np.meshgrid(r, theta)

X = r*np.sin(theta)
Y = r*np.cos(theta)
Z = f(X,Y)
```

```python
fig = plt.figure(figsize = (12, 8))

ax = plt.axes(projection = '3d')

ax.plot_surface(X, Y, Z, cmap = 'summer')

plt.show()
```
